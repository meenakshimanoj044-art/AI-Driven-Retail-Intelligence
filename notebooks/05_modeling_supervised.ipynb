```python
import pandas as pd
import numpy as np
from pathlib import Path
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor



```

```python
from lightgbm import LGBMRegressor
print("WORKING")
```

```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path().resolve().parent   # adjust parent level if needed
dataset_path = BASE_DIR / "data/processed/regression_features.csv"

df = pd.read_csv(dataset_path)
```

```python
'''. Predict Next Month Product Sales
Business Question:
Can we predict the quantity of products sold in the next
month using historical
sales trends, pricing patterns, and seasonal effects?
Target Variable:
Monthly product sales quantity
Business Value:
Improves inventory planning and demand forecasting.'''




```

```python
df["target"] = df.groupby("StockCode")["MonthlyQuantity"].shift(-1)
df = df.dropna(subset=["target"])
```

```python
train = df[df["YearMonth"] <= "2011-10"]
test = df[df["YearMonth"] > "2011-10"]
```

```python
X_train = train.drop(columns=["target"])
y_train = train["target"]
X_test = test.drop(columns=["target"])
y_test = test["target"]
```

```python
X_train = pd.get_dummies(X_train)
X_test = pd.get_dummies(X_test)

X_train, X_test = X_train.align(X_test, join="left", axis=1, fill_value=0)
```

```python
models={
    "LinearRegression":LinearRegression(),
    "Lasso":Lasso(),
    "Ridge":Ridge(),
    "Decision Tree":DecisionTreeRegressor(),
    "Random Forest":RandomForestRegressor(),
    "LightGBM":LGBMRegressor()

}
```

```python
results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test,predictions))

    results[name] = {"MAE": mae, "RMSE": rmse}

    print(f"{name}")
    print(f"MAE: {mae:.3f}")
    print(f"RMSE: {rmse:.3f}")
    print("-" * 30)
```

```python
best_model = min(results, key=lambda x: results[x]["RMSE"])

print("BEST MODEL:", best_model)
print("Performance:", results[best_model])
```
