```python
🚀 4. BASIC LLM SYNTAX (GENERIC TEMPLATE)
from openai import OpenAI

client = OpenAI(api_key="YOUR_KEY")

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are an expert assistant"},
        {"role": "user", "content": "Explain clustering"}
    ]
)

print(response.choices[0].message.content)
🧠 5. WHAT EACH LINE MEANS
🔹 model
model="gpt-4o-mini"

👉 which AI brain to use

🔹 messages
messages=[...]

👉 conversation history (context)

🔹 role + content
{"role": "user", "content": "..."}

👉 structured communication

🔹 response extraction
response.choices[0].message.content
```

Let’s break it down simply:

1. OpenAI(...)

This is the official OpenAI Python SDK class.

It creates a client (like a bridge)
Through this client, you can send requests like:
chat completions
embeddings
images, etc.

So instead of manually calling HTTP requests, you use this object.

2. api_key=...

This is where your secret key goes.

The API key is what proves: “you are allowed to use OpenAI services”
Without it, the API will reject your request
3. os.getenv("OPENAI_API_KEY")

This part fetches the key from your environment variables.

os → Python module for interacting with system environment
getenv() → “get environment variable”
"OPENAI_API_KEY" → the name of the variable you stored earlier

So it is basically saying:

“Go and find my API key from the system settings, don’t hardcode it here.”
