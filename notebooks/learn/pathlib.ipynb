
1. ``from pathlib import Path: Imports the modern Python pathlib module, which treats file paths as objects rather than simple strings.``
2. ``import pandas as pd: Imports the pandas library, used for data manipulation and analysis.``

3. ``Path(__file__): Identifies the location of the current script file being executed.``

4. ``.resolve(): Converts that location into an absolute path and follows any symbolic links to find the "real" physical location on the disk.``

5. ``.parent.parent: Navigates two levels up the directory tree from the current file.``


Why use this?
Portability: Your code won't break if you move the project folder to a different drive or server, because it calculates the path relative to the code itself.
Clean Syntax: Instead of complex string joins, you can find files easily using the / operator, like DATA_FILE = BASE_DIR / "data" / "file.csv".

```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path(__file__).resolve().parent.parent

def load_data():
    retail_path = BASE_DIR / "data/raw/online_retail.csv"
    holidays_path = BASE_DIR / "data/external/publicHolidays.csv"

    raw_online_retail = pd.read_csv(retail_path)
    raw_public_holidays = pd.read_csv(holidays_path)
```
